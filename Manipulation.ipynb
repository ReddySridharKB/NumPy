```python
import numpy as np

arr = np.array([1, 2, 3, 4, 5, 6])

new_arr = arr.reshape(2, 3)

print(new_arr)
```

```python
arr = np.array([[1, 2, 3],
                [4, 5, 6]])

new_arr = arr.flatten()

print(new_arr)
```

```python
arr = np.array([[1, 2, 3],
                [4, 5, 6]])

new_arr = arr.ravel()

print(new_arr)
```

```python
arr = np.array([[1, 2, 3],
                [4, 5, 6]])

print(arr.T)
```

```python
arr1 = np.array([1, 2, 3])
arr2 = np.array([4, 5, 6])

result = np.concatenate((arr1, arr2))

print(result)
```

```python
arr1 = np.array([[1, 2],
                 [3, 4]])

arr2 = np.array([[5, 6],
                 [7, 8]])

result = np.concatenate((arr1, arr2), axis=0)

print(result)
```
